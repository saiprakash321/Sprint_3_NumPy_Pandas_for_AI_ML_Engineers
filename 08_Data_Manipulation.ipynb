## 1. Rename Columns

Renaming Columns means changing the existing column names of a Pandas DataFrame.

The `rename()` method can be used to give columns clearer or more meaningful names.

### Real-World Example

A student dataset may contain a column named `Marks`. We can rename it to `Score` to make the column name more suitable for a particular analysis.

### Business Example

A sales dataset may contain `Amt`. It can be renamed to `Sales_Amount` to make the data easier to understand.

### AI/ML Example

During data preprocessing, meaningful column names make it easier to identify and work with features.

```python
import pandas as pd

data = {
    "Name": ["Sai", "Rahul", "Priya"],
    "Age": [23, 22, 20],
    "Marks": [85, 78, 92]
}

df = pd.DataFrame(data)

df = df.rename(columns={"Marks": "Score"})

print(df)
```

### Explanation

The `rename()` method changes the column name `Marks` to `Score`.

The values inside the column remain unchanged.

### Best Practices

Use clear and meaningful column names that describe the data.

### Common Mistakes

Using a column name that does not exist will not rename the expected column.

## 2. Drop Rows

Dropping Rows means removing selected rows from a DataFrame.

The `drop()` method can be used to remove rows using their index labels.

### Real-World Example

A student record that is no longer required can be removed from a dataset.

### Business Example

A company can remove an incorrect or unwanted customer record from a dataset.

### AI/ML Example

During data preprocessing, unwanted or invalid records can be removed before training a Machine Learning model.

```python
data = {
    "Name": ["Sai", "Rahul", "Priya"],
    "Age": [23, 22, 20],
    "Marks": [85, 78, 92]
}

df = pd.DataFrame(data)

df = df.drop(1)

print(df)
```

### Explanation

The `drop(1)` method removes the row with index `1`.

The remaining rows keep their original index labels.

### Best Practices

Check the row index before removing rows.

### Common Mistakes

Dropping the wrong index can remove an unwanted record.

## 3. Drop Columns

Dropping Columns means removing one or more columns from a DataFrame.

The `drop()` method can be used with `axis=1` to remove columns.

### Real-World Example

If a student dataset contains unnecessary information, an unused column can be removed.

### Business Example

A company may remove columns that are not required for a particular sales analysis.

### AI/ML Example

Unnecessary features can be removed during data preprocessing to create a more suitable dataset for Machine Learning.

```python
data = {
    "Name": ["Sai", "Rahul", "Priya"],
    "Age": [23, 22, 20],
    "Marks": [85, 78, 92]
}

df = pd.DataFrame(data)

df = df.drop("Age", axis=1)

print(df)
```

### Explanation

The `drop()` method removes the `Age` column.

The `axis=1` tells Pandas that we want to remove a column.

### Best Practices

Remove only columns that are unnecessary for the required analysis.

### Common Mistakes

Using `axis=0` when trying to remove a column will target rows instead.

## 4. Insert Columns

Inserting a Column means adding a new column at a specific position in a DataFrame.

The `insert()` method allows us to specify where the new column should be placed.

### Real-World Example

A student dataset can have a new `Grade` column added after the Marks column.

### Business Example

A company can add a calculated category or status column to an existing sales dataset.

### AI/ML Example

A new feature can be added to a dataset during data preparation and feature engineering.

```python
data = {
    "Name": ["Sai", "Rahul", "Priya"],
    "Marks": [85, 78, 92]
}

df = pd.DataFrame(data)

df.insert(2, "Age", [23, 22, 20])

print(df)
```

### Explanation

The `insert()` method adds the `Age` column at position `2`.

The existing columns are shifted to make space for the new column.

### Best Practices

Make sure the position and number of values match the DataFrame correctly.

### Common Mistakes

Using an invalid column position or providing a different number of values can cause an error.

## 5. assign()

The `assign()` method is used to create new columns or modify existing columns in a DataFrame.

It returns a new DataFrame with the assigned column.

### Real-World Example

A student dataset can use `assign()` to create a new column based on existing marks.

### Business Example

A company can create a new revenue-related column using existing sales information.

### AI/ML Example

`assign()` can be useful during feature engineering to create new features from existing data.

```python
data = {
    "Name": ["Sai", "Rahul", "Priya"],
    "Marks": [85, 78, 92]
}

df = pd.DataFrame(data)

new_df = df.assign(Passed=df["Marks"] >= 40)

print(new_df)
```

### Explanation

The `assign()` method creates a new column named `Passed`.

The value is `True` when the student's marks are greater than or equal to 40.

### Best Practices

Use `assign()` when creating new columns while keeping the original DataFrame unchanged.

### Common Mistakes

Forgetting that `assign()` returns a new DataFrame when the result is not stored.

## 6. apply()

The `apply()` method is used to apply a function to values in a Pandas Series or DataFrame.

It is useful when we need to perform the same operation on multiple values.

### Real-World Example

A student's marks can be processed to determine whether the student passed or failed.

### Business Example

A company can apply a calculation to sales values to create categories or perform transformations.

### AI/ML Example

`apply()` can be used during data preprocessing to transform values or create useful features.

```python
data = {
    "Name": ["Sai", "Rahul", "Priya"],
    "Marks": [85, 38, 92]
}

df = pd.DataFrame(data)

df["Result"] = df["Marks"].apply(
    lambda x: "Pass" if x >= 40 else "Fail"
)

print(df)
```

### Explanation

The `apply()` method applies the given function to each value in the Marks column.

Students with marks of 40 or above are assigned `Pass`, while others are assigned `Fail`.

### Best Practices

Use `apply()` when a custom operation is required that cannot be easily performed with a standard Pandas operation.

### Common Mistakes

Using `apply()` unnecessarily for simple operations can make code slower and harder to understand.

## 7. map()

The `map()` method is used to transform values in a Pandas Series.

It is commonly used to replace existing values with corresponding values from a mapping.

### Real-World Example

Student marks can be converted into grades using a mapping between marks and grades.

### Business Example

A company can convert short product codes into their full product names.

### AI/ML Example

Categorical values can be converted into new labels during data preprocessing.

```python
data = {
    "Name": ["Sai", "Rahul", "Priya"],
    "Grade": ["A", "B", "A"]
}

df = pd.DataFrame(data)

grade_map = {
    "A": "Excellent",
    "B": "Good"
}

df["Performance"] = df["Grade"].map(grade_map)

print(df)
```

### Explanation

The `map()` method replaces each grade with its corresponding value from `grade_map`.

`A` becomes `Excellent` and `B` becomes `Good`.

### Best Practices

Make sure every value that needs to be mapped has a corresponding mapping value.

### Common Mistakes

Values that are not present in the mapping dictionary may become `NaN`.

## 8. replace()

The `replace()` method is used to replace specific values in a Pandas DataFrame or Series.

It is useful when existing values need to be corrected or changed.

### Real-World Example

A dataset may contain different spellings or incorrect values that need to be replaced.

### Business Example

A company can replace short category codes with standardized values.

### AI/ML Example

During data cleaning, incorrect or inconsistent values can be replaced with appropriate values before Machine Learning.

```python
data = {
    "Name": ["Sai", "Rahul", "Priya"],
    "Status": ["Yes", "No", "Yes"]
}

df = pd.DataFrame(data)

df["Status"] = df["Status"].replace({
    "Yes": "Active",
    "No": "Inactive"
})

print(df)
```

### Explanation

The `replace()` method changes `Yes` to `Active` and `No` to `Inactive`.

It is useful for correcting or standardizing values in a dataset.

### Best Practices

Use `replace()` when specific values need to be changed or standardized.

### Common Mistakes

Replacing values without checking the original data can lead to unintended changes.

## 9. Duplicate Handling

Duplicate Handling is the process of identifying and managing repeated rows in a DataFrame.

Pandas provides methods such as `duplicated()` and `drop_duplicates()` for working with duplicate records.

### Real-World Example

A student record may accidentally appear more than once in a dataset. Duplicate records can be identified and removed.

### Business Example

A customer may be entered into a company's database multiple times. Duplicate handling helps maintain cleaner customer data.

### AI/ML Example

Duplicate records can affect data analysis and Machine Learning results. Removing unnecessary duplicates can help improve data quality.

```python
data = {
    "Name": ["Sai", "Rahul", "Sai", "Priya"],
    "Age": [23, 22, 23, 20],
    "Marks": [85, 78, 85, 92]
}

df = pd.DataFrame(data)

print(df.duplicated())
```

### Explanation

The `duplicated()` method checks whether each row is a duplicate of an earlier row.

The third row is marked as `True` because the same record for Sai already exists.

### Best Practices

Check duplicate records before removing them to make sure they are actually unwanted duplicates.

### Common Mistakes

Removing duplicates without checking whether repeated records are valid can result in loss of useful information.

```python
df = df.drop_duplicates()

print(df)
```

### Explanation

The `drop_duplicates()` method removes the repeated record from the DataFrame.

The remaining rows contain only unique records.

### Best Practices

Always inspect duplicate records before permanently removing them.

### Common Mistakes

Assuming every repeated row is an error. Some datasets may legitimately contain repeated records.
