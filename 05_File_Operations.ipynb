## 1. CSV Files

CSV stands for Comma-Separated Values. It is a simple file format used to store tabular data in rows and columns.

CSV files are commonly used because they are easy to create, read, and share. Pandas provides functions such as `read_csv()` and `to_csv()` for working with CSV files.

### Real-World Example

Student information such as names, ages, and marks can be stored in a CSV file and loaded into Pandas for analysis.

### Business Example

A company can store customer or sales information in a CSV file and use Pandas to analyze the data.

### AI/ML Example

Many Machine Learning datasets are available as CSV files. Pandas can load these datasets into a DataFrame before data cleaning and preprocessing.

```python
import pandas as pd

df = pd.read_csv("students.csv")

print(df)
```

###  Explanation

The `pd.read_csv()` function reads the CSV file and converts its data into a Pandas DataFrame.

The DataFrame contains the Name, Age, and Marks columns.

### Best Practices

Check the file name and file path before reading a CSV file.

### Common Mistakes

Using an incorrect file name or file path can cause a `FileNotFoundError`.

## 2. Excel Files

Excel files are commonly used to store data in rows and columns. Pandas provides functions such as `read_excel()` and `to_excel()` to read and write Excel files.

Excel files are useful when working with structured data that needs to be viewed or managed using spreadsheet applications.

### Real-World Example

Student records such as names, ages, marks, and grades can be stored in an Excel file and loaded into Pandas for analysis.

### Business Example

A company can store sales, customer, or employee information in an Excel file and use Pandas to analyze the data.

### AI/ML Example

Excel datasets can be loaded into Pandas DataFrames and then cleaned and prepared before using the data for Machine Learning.

```python
%pip install openpyxl
```

```python
import pandas as pd

data = {
    "Name": ["Saiprakash", "Neha", "Sampradaya"],
    "Age": [23, 22, 20],
    "Marks": [85, 78, 92]
}

df = pd.DataFrame(data)

df.to_excel("students.xlsx", index=False)

print("Excel file created successfully")
```

```python
df = pd.read_excel("students.xlsx")

print(df)
```

###  Explanation

The `to_excel()` function saves the DataFrame as an Excel file.

The `read_excel()` function reads the Excel file and loads the data into a DataFrame.

### Best Practices

Use `index=False` when the DataFrame index is not required in the Excel file.

### Common Mistakes

Using an incorrect file name or path, or not having the required Excel engine installed.

## 3. JSON Files

JSON stands for JavaScript Object Notation. It is a common file format used to store and exchange structured data.

JSON files store data using key-value pairs and can represent structured information such as objects, lists, and records.

Pandas provides functions such as `read_json()` and `to_json()` to read and write JSON data.

### Real-World Example

Product information such as product names, prices, and categories can be stored in a JSON file and loaded into Pandas for analysis.

### Business Example

A company can use JSON files to store customer, product, or transaction data exchanged between different applications.

### AI/ML Example

JSON files can contain structured datasets or API responses that can be loaded into Pandas and prepared for Machine Learning tasks.

```python
import pandas as pd

data = {
    "Name": ["Saiprakash", "Neha", "Sampradaya"],
    "Age": [23, 22, 20],
    "Marks": [85, 78, 92]
}

df = pd.DataFrame(data)

df.to_json("students.json", orient="records")

print("JSON file created successfully")
```

```python
df = pd.read_json("students.json")

print(df)
```

###  Explanation

The `to_json()` function saves the DataFrame as a JSON file.

The `read_json()` function reads the JSON file and converts the data into a Pandas DataFrame.

The `orient="records"` option stores each row as a separate JSON object.

### Best Practices

Use a suitable JSON orientation depending on how the data needs to be stored or shared.

### Common Mistakes

Using an incorrect JSON file path or using an unsuitable JSON structure for the required operation.

## 4. Parquet Files

Parquet is a column-based file format commonly used for storing and processing large datasets.

It stores data efficiently and can provide faster reading and writing compared with some traditional file formats.

Pandas provides `read_parquet()` and `to_parquet()` for working with Parquet files.

### Real-World Example

A large dataset containing customer or transaction records can be stored in Parquet format for efficient storage and processing.

### Business Example

Companies can store large sales and customer datasets in Parquet files and use them for data analysis and reporting.

### AI/ML Example

Large Machine Learning datasets can be stored in Parquet format and loaded into Pandas for data preparation and model development.

## 5. Pickle Files

Pickle is a Python file format used to store Python objects in a serialized form.

Pandas provides `to_pickle()` to save a DataFrame and `read_pickle()` to load it back into Python.

Pickle is useful when we want to save and restore Python data without converting it into a text-based format.

### Real-World Example

A DataFrame containing student records can be saved as a Pickle file and loaded later without creating the DataFrame again.

### Business Example

A company can save a processed DataFrame as a Pickle file and reuse it later during data analysis.

### AI/ML Example

Processed datasets or Python objects can be stored using Pickle and loaded again when continuing Machine Learning experiments.

```python
import pandas as pd

data = {
    "Name": ["Saiprakash", "N", "Sam"],
    "Age": [23, 22, 20],
    "Marks": [85, 78, 92]
}

df = pd.DataFrame(data)

df.to_pickle("students.pkl")

print("Pickle file created successfully")
```

```python
df = pd.read_pickle("students.pkl")

print(df)
```

###  Explanation

The `to_pickle()` function saves the DataFrame as a Pickle file.

The `read_pickle()` function loads the saved Pickle file back into a Pandas DataFrame.

### Best Practices

Use Pickle mainly when working within a trusted Python environment and when you need to preserve Python objects.

### Common Mistakes

Do not load Pickle files from untrusted sources because Pickle files can execute arbitrary code when loaded.

## 6. SQL Tables

SQL tables are used to store structured data in relational databases.

Pandas can read data from SQL tables and convert it into a DataFrame. It can also write a DataFrame into a SQL table.

This is useful when data is stored in a database instead of a file.

### Real-World Example

Student information can be stored in a database table and loaded into Pandas for analysis.

### Business Example

A company can store customer, sales, and transaction data in a SQL database and use Pandas to analyze the data.

### AI/ML Example

Machine Learning data can be retrieved from SQL databases into Pandas DataFrames before cleaning and preprocessing.

```python
import pandas as pd
import sqlite3

data = {
    "Name": ["Sai", "NNN", "Sam"],
    "Age": [25, 28, 22],
    "Marks": [85, 78, 92]
}

df = pd.DataFrame(data)

connection = sqlite3.connect("students.db")

df.to_sql("students", connection, if_exists="replace", index=False)

print("Data saved to SQL table")
```

```python
df = pd.read_sql("SELECT * FROM students", connection)

print(df)
```

```python
connection.close()

print("Database connection closed")
```

###  Explanation

The `to_sql()` function saves the DataFrame into a SQL table.

The `read_sql()` function reads data from the SQL table and converts it into a DataFrame.

SQLite is used here because it provides a simple database without requiring a separate database server.

### Best Practices

Close the database connection after completing the database operations.

### Common Mistakes

Using an incorrect table name or SQL query can cause an error when reading data.

## 7. Reading Multiple Files

Sometimes data is stored in multiple files instead of a single file.

Pandas can be combined with Python's `glob` module to find and read multiple files from a folder.

The data from multiple files can then be combined into a single DataFrame for analysis.

### Real-World Example

Monthly student or attendance records may be stored in separate CSV files. These files can be read together and combined for analysis.

### Business Example

A company may store sales data in separate files for different months or regions. Multiple files can be combined to analyze the complete sales data.

### AI/ML Example

Machine Learning datasets may be divided into multiple files. Reading and combining these files can help create a single dataset for preprocessing and model training.

```python
import pandas as pd
import glob

files = glob.glob("*.csv")

data = []

for file in files:
    df = pd.read_csv(file)
    data.append(df)

combined_df = pd.concat(data, ignore_index=True)

print(combined_df)
```

### Explanation

The `glob.glob()` function finds CSV files in the current folder.

Each file is read using `pd.read_csv()` and stored in a list.

The `pd.concat()` function combines the DataFrames into one DataFrame.

### Best Practices

Make sure the files have compatible columns before combining them.

### Common Mistakes

Trying to combine files with different column structures can produce unexpected results.

## 8. File Path Handling

File paths are used to identify the location of files and folders on a computer.

When working with Pandas, providing the correct file path is important for reading and writing data.

Python provides the `pathlib` module to work with file and folder paths in a simple and reliable way.

### Real-World Example

A CSV file may be stored inside a specific folder on a computer. The correct file path is required to read that file using Pandas.

### Business Example

A company may organize sales, customer, and product files into different folders. File paths help applications locate and process these files.

### AI/ML Example

Machine Learning projects often contain separate folders for datasets, models, and results. Proper file path handling helps load and save these files correctly.

```python
from pathlib import Path

file_path = Path("students.csv")

print("File name:", file_path.name)
print("File exists:", file_path.exists())
```

###  Explanation

The `Path()` function creates a file path object.

The `name` attribute returns the file name.

The `exists()` method checks whether the file is available at the given path.

### Best Practices

Use `pathlib` for handling file paths because it makes path operations easier and more reliable across different operating systems.

### Common Mistakes

Using an incorrect file path or file name can cause a `FileNotFoundError`.
