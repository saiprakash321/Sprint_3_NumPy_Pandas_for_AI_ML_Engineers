## 1. lower()

The `lower()` method converts all uppercase letters in a string to lowercase letters.

It is useful for making text values consistent when working with string data.

### Real-World Example

Names entered by users may contain different combinations of uppercase and lowercase letters. Converting them to lowercase helps keep the data consistent.

### Business Example

A company can convert customer names, email addresses, or product categories to lowercase before comparing or analyzing them.

### AI/ML Example

In text preprocessing, converting text to lowercase helps reduce differences between words such as "Python", "PYTHON", and "python".

```python
import pandas as pd

data = {
    "Name": ["Sai", "RAHUL", "Priya"]
}

df = pd.DataFrame(data)

df["Name"] = df["Name"].str.lower()

print(df)
```

### Explanation

The `str.lower()` operation converts every letter in the Name column to lowercase.

### Best Practices

Use lowercase conversion when text needs to be compared or standardized.

### Common Mistakes

Forgetting to use `.str` when applying string methods to a Pandas Series.

## 2. upper()

The `upper()` method converts all lowercase letters in a string to uppercase letters.

It is useful when text needs to follow a consistent uppercase format.

### Real-World Example

Names, codes, or labels entered in different formats can be converted to uppercase for consistency.

### Business Example

A company can convert product codes or department names to uppercase before storing or comparing them.

### AI/ML Example

Text preprocessing can use uppercase or lowercase conversion to standardize text before further analysis.

```python
data = {
    "Name": ["Sai", "rahul", "Priya"]
}

df = pd.DataFrame(data)

df["Name"] = df["Name"].str.upper()

print(df)
```

### Explanation

The `str.upper()` operation converts all letters in the Name column to uppercase.

### Best Practices

Use uppercase conversion when the data requires a consistent uppercase format.

### Common Mistakes

Assuming `upper()` changes numbers or special characters. It only affects alphabetic characters.

## 3. strip()

The `strip()` method removes unwanted spaces or specified characters from the beginning and end of a string.

It is commonly used to clean text data.

### Real-World Example

A student's name may contain accidental spaces before or after the name.

### Business Example

Customer or product information collected from forms may contain extra spaces that should be removed before analysis.

### AI/ML Example

During text preprocessing, removing unnecessary spaces helps create cleaner and more consistent text data.

```python
data = {
    "Name": [" Sai ", "Rahul ", " Priya"]
}

df = pd.DataFrame(data)

df["Name"] = df["Name"].str.strip()

print(df)
```

### Explanation

The `str.strip()` operation removes leading and trailing spaces from the Name column.

### Best Practices

Use `strip()` when data may contain unwanted spaces around text values.

### Common Mistakes

Expecting `strip()` to remove spaces between words. It only removes characters from the beginning and end.

## 4. split()

The `split()` method divides a string into smaller parts based on a specified separator.

It is useful when one string contains multiple pieces of information.

### Real-World Example

A full name such as "Sai Kumar" can be split into separate parts.

### Business Example

A company can split product codes, addresses, or category values into separate components.

### AI/ML Example

Text can be split into words or smaller parts during Natural Language Processing and text preprocessing.

```python
data = {
    "Name": ["Sai Kumar", "Rahul Sharma", "Priya Singh"]
}

df = pd.DataFrame(data)

df["First_Name"] = df["Name"].str.split().str[0]

print(df)
```

### Explanation

The `str.split()` operation separates each full name into individual parts.

`str[0]` selects the first part of each name and stores it in the `First_Name` column.

### Best Practices

Choose the correct separator when splitting text.

### Common Mistakes

Assuming every string has the same number of parts can cause unexpected results.

## 5. contains()

The `contains()` method checks whether a string contains a specific word, character, or pattern.

It returns `True` when the value is found and `False` when it is not.

### Real-World Example

A list of student names can be searched to find names containing a particular letter or word.

### Business Example

A company can search customer feedback to find records containing words such as "good", "late", or "delivery".

### AI/ML Example

In text preprocessing, `contains()` can help identify records containing particular keywords or patterns.

```python
data = {
    "Name": ["Sai Kumar", "Rahul Sharma", "Priya Singh"]
}

df = pd.DataFrame(data)

result = df[df["Name"].str.contains("Sai")]

print(result)
```

### Explanation

The `str.contains("Sai")` checks whether the Name column contains the text `Sai`.

Only the row containing `Sai` is returned.

### Best Practices

Use `contains()` when searching text for specific words or patterns.

### Common Mistakes

Text matching can be case-sensitive unless case handling is specified.

## 6. replace()

The `replace()` method is used to replace one piece of text with another.

It is useful for correcting, cleaning, or standardizing string values.

### Real-World Example

A student's name may contain an incorrect spelling that needs to be corrected.

### Business Example

A company can replace old product names or inconsistent category labels with standardized values.

### AI/ML Example

Text preprocessing can use `replace()` to remove unwanted characters or standardize words before analysis.

```python
data = {
    "Name": ["Sai Kumar", "Rahul Sharma", "Priya Singh"]
}

df = pd.DataFrame(data)

df["Name"] = df["Name"].str.replace("Kumar", "K.", regex=False)

print(df)
```

### Explanation

The `str.replace()` operation changes `Kumar` to `K.` in the Name column.

The other names remain unchanged.

### Best Practices

Check the original data before replacing text to make sure the replacement is appropriate.

### Common Mistakes

Using an incorrect replacement value can change data unintentionally.

## 7. Regular Expressions (Regex)

Regular Expression, commonly called Regex, is a pattern used to search, match, or extract specific text from a string.

Regex is useful when simple string methods are not enough to identify a particular pattern.

### Real-World Example

A phone number or email address can be checked to see whether it follows a particular pattern.

### Business Example

A company can use Regex to extract phone numbers, email addresses, product codes, or other structured information from text.

### AI/ML Example

Regex can be used during text preprocessing to remove unwanted characters, extract useful patterns, and clean text before Natural Language Processing tasks.

```python
import re

text = "Contact Sai at sai123@gmail.com"

email = re.findall(
    r"[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}",
    text
)

print(email)
```

### Explanation

The Regex pattern searches the text for an email address.

`re.findall()` returns all matching email addresses found in the text.

### Best Practices

Use simple and clear Regex patterns and test them with different types of input.

### Common Mistakes

A Regex pattern that is too simple or too complex may fail to match the required text correctly.
