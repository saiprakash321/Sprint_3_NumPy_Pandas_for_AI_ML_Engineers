### 3. Explain Broadcasting in NumPy with a simple example.

### Answer

Broadcasting is a NumPy feature that allows arithmetic operations between arrays of different but compatible shapes.

For example, a single value can be added to every element of a NumPy array without using a loop.

```python
import numpy as np

data = np.array([10, 20, 30])

result = data + 5

print("Original Array:", data)
print("After Broadcasting:", result)
```

### 6. Merge Two DataFrames

### Answer

Create two Pandas DataFrames with a common column and merge them using `merge()`.



```python
import pandas as pd

students = pd.DataFrame({
    "Student_ID": [1, 2, 3],
    "Name": ["Sai", "Rahul", "Priya"]
})

marks = pd.DataFrame({
    "Student_ID": [1, 2, 3],
    "Marks": [85, 78, 92]
})

result = pd.merge(students, marks, on="Student_ID")

print(result)
```

### 7. Perform GroupBy and Aggregation


### Answer

Create a Pandas DataFrame containing a grouping column and a numerical column.

Use `groupby()` with an aggregation function to calculate the average marks for each department.



```python
import pandas as pd

data = {
    "Name": ["Sai", "Rahul", "Priya", "Amit"],
    "Department": ["IT", "HR", "IT", "HR"],
    "Marks": [85, 78, 92, 88]
}

df = pd.DataFrame(data)

result = df.groupby("Department")["Marks"].mean()

print(result)
```

### 8. Filter Rows Based on Multiple Conditions

### Answer

Create a Pandas DataFrame and filter students whose marks are greater than 80 and age is less than 23

```python
import pandas as pd

data = {
    "Name": ["Sai", "Rahul", "Priya", "Amit"],
    "Age": [23, 22, 20, 21],
    "Marks": [85, 78, 92, 88]
}

df = pd.DataFrame(data)

result = df[(df["Marks"] > 80) & (df["Age"] < 23)]

print(result)
```

### 9. Handle Missing Values

### Answer

Create a Pandas DataFrame containing missing values.

Identify the missing values and handle them using an appropriate Pandas method.



```python
import pandas as pd

df = pd.DataFrame({
    "Name": ["Sai", "Rahul", "Priya"],
    "Age": [23, None, 20],
    "Marks": [85, 78, None]
})

print(df.isnull().sum())

df["Age"] = df["Age"].fillna(df["Age"].mean())
df["Marks"] = df["Marks"].fillna(df["Marks"].mean())

print(df)
```

### 10. Statistical Measures Using NumPy

### Answer

Create a NumPy array and calculate the sum, mean, median, variance, standard deviation, minimum, and maximum.



```python
import numpy as np

data = np.array([10, 20, 30, 40, 50])

print("Sum:", np.sum(data))
print("Mean:", np.mean(data))
print("Median:", np.median(data))
print("Variance:", np.var(data))
print("Standard Deviation:", np.std(data))
print("Minimum:", np.min(data))
print("Maximum:", np.max(data))
```

### 11. Optimize a DataFrame for Memory Usage


### Answer

Create a Pandas DataFrame containing numerical and categorical data.

Check the memory usage and optimize the DataFrame using a suitable data type.



```python
import pandas as pd

df = pd.DataFrame({
    "Name": ["Sai", "Rahul", "Priya", "Amit"],
    "Department": ["IT", "HR", "IT", "HR"],
    "Marks": [85, 78, 92, 88]
})

print("Before:")
print(df.memory_usage(deep=True))

df["Department"] = df["Department"].astype("category")

print("\nAfter:")
print(df.memory_usage(deep=True))
```
