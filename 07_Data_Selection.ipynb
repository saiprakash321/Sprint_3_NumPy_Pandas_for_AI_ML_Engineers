### Module 7 – Selecting & Filtering Data



## 1. loc – Location

`loc` stands for **Location**. It is a Pandas DataFrame accessor used to select rows and columns using their labels.

It helps us access specific rows, columns, or values from a DataFrame using row labels and column names.

### Real-World Example

A student dataset can contain many student records. Using `loc`, we can select a particular student's information or specific columns such as Name and Marks.

### Business Example

A company can use `loc` to select specific customer records or columns such as Customer Name, Location, and Revenue.

### AI/ML Example

During data preprocessing, `loc` can be used to select required rows and features from a dataset before applying Machine Learning algorithms.

```python
import pandas as pd

data = {
    "Name": ["Sai", "Rahul", "Priya", "Amit"],
    "Age": [23, 22, 20, 21],
    "Marks": [85, 78, 92, 88]
}

df = pd.DataFrame(data)

print(df.loc[0])
```

### Explanation

`df.loc[0]` selects the row with label `0`.

We can also use `loc` to select specific values, rows, and columns from the DataFrame.

### Best Practices

Use `loc` when selecting data based on row labels and column names.

### Common Mistakes

Do not confuse `loc` with `iloc`. `loc` uses labels, while `iloc` uses integer positions.

## 2. iloc – Integer Location

`iloc` stands for **Integer Location**. It is a Pandas DataFrame accessor used to select rows and columns based on their integer positions.

It is useful when we want to access data based on the position of a row or column rather than its label.

### Real-World Example

A student dataset can be accessed by selecting the first student or specific columns based on their positions.

### Business Example

A company can use `iloc` to select records or columns from a customer or sales dataset based on their positions.

### AI/ML Example

During data preprocessing, `iloc` can be used to select specific rows or features based on their positions.


```python
print(df.iloc[0])
```

### Explanation

`df.iloc[0]` selects the first row of the DataFrame because its position is `0`.

Unlike `loc`, `iloc` works with integer positions.

### Best Practices

Use `iloc` when you need to select data based on its position.

### Common Mistakes

Do not confuse integer position with the row label. `iloc` uses positions, while `loc` uses labels.

## 3. Boolean Indexing

Boolean Indexing is a technique used to filter rows in a Pandas DataFrame based on a condition.

It returns only the rows where the given condition is `True`.

### Real-World Example

A student dataset can be filtered to find students who scored more than 80 marks.

### Business Example

A company can use Boolean Indexing to find customers whose purchase amount is greater than a specific value.

### AI/ML Example

During data preprocessing, Boolean Indexing can be used to select records that satisfy specific conditions before training a Machine Learning model.

```python
filtered_df = df[df["Marks"] > 80]

print(filtered_df)
```

### Explanation

The condition `df["Marks"] > 80` checks each value in the Marks column.

Only the rows where the marks are greater than 80 are returned.

### Best Practices

Use clear conditions when filtering DataFrames and check the resulting data after filtering.

### Common Mistakes

Using a single `=` instead of `==` when comparing values can cause an error.

## 4. Filtering Rows

Filtering Rows is the process of selecting only the rows that meet a specific condition.

It helps us focus on the records that are relevant to a particular requirement.

### Real-World Example

A student dataset can be filtered to find students whose age is greater than 20.

### Business Example

A company can filter customer records to find customers from a particular location or customers with high purchase amounts.

### AI/ML Example

During data preprocessing, filtering can be used to select relevant records and remove data that does not meet specific conditions.

```python
filtered_df = df[df["Age"] > 20]

print(filtered_df)
```

### Explanation

The condition `df["Age"] > 20` checks the Age column.

Only the rows where the age is greater than 20 are selected.

### Best Practices

Use clear conditions and verify the filtered DataFrame before using it for further analysis.

### Common Mistakes

Using the wrong comparison operator can return incorrect rows.

## 5. query()

`query()` is a Pandas DataFrame method used to filter rows by writing a condition as a query expression.

It makes filtering easier to read when working with one or more conditions.

### Real-World Example

A student dataset can be filtered to find students whose marks are greater than 80.

### Business Example

A company can use `query()` to find customers whose purchase amount is above a specific value.

### AI/ML Example

During data preprocessing, `query()` can be used to select records that satisfy specific conditions before further analysis or Machine Learning.


```python
filtered_df = df.query("Marks > 80")

print(filtered_df)
```

### Explanation

The `query()` method filters the DataFrame using the condition `"Marks > 80"`.

Only the students whose marks are greater than 80 are returned.

### Best Practices

Use `query()` when a filtering condition is easier to understand as a readable expression.

### Common Mistakes

Using an incorrect column name or writing the query condition incorrectly can cause an error.

## 6. isin()

`isin()` is a Pandas method used to check whether values in a column are present in a given list of values.

It is useful when we want to filter rows that match one or more specific values.

### Real-World Example

A student dataset can be filtered to select students whose names are present in a particular list.

### Business Example

A company can use `isin()` to select customers from specific cities or products from selected categories.

### AI/ML Example

During data preprocessing, `isin()` can be used to select records belonging to specific categories or classes.


```python
selected_students = df[df["Name"].isin(["Sai", "Priya"])]

print(selected_students)
```

### Explanation

The `isin()` method checks whether each value in the Name column matches one of the names in the given list.

Only the rows containing `Sai` or `Priya` are selected.

### Best Practices

Use `isin()` when filtering a column against multiple specific values.

### Common Mistakes

Using values that do not exactly match the values in the column can result in no rows being selected.

## 7. between()

`between()` is a Pandas method used to check whether values in a column fall within a specified range.

It is useful when we want to filter data between a minimum and maximum value.

### Real-World Example

A student dataset can be filtered to find students whose marks are between 80 and 90.

### Business Example

A company can use `between()` to find products whose prices fall within a particular price range.

### AI/ML Example

During data preprocessing, `between()` can be used to select records whose numerical values fall within a required range.

```python
filtered_df = df[df["Marks"].between(80, 90)]

print(filtered_df)
```

### Explanation

The `between(80, 90)` method checks whether the Marks value is between 80 and 90.

Only the students whose marks fall within this range are selected.

### Best Practices

Use `between()` when filtering numerical values within a specific range.

### Common Mistakes

Using an incorrect minimum or maximum value can return unexpected results.

## 8. Sorting Data

Sorting Data is the process of arranging rows in a DataFrame based on the values of one or more columns.

Pandas provides the `sort_values()` method to arrange data in ascending or descending order.

### Real-World Example

A student dataset can be sorted according to marks to identify students with the highest or lowest scores.

### Business Example

A company can sort sales records by revenue to identify the highest or lowest performing products.

### AI/ML Example

During data analysis, sorting can help understand patterns and examine numerical values before Machine Learning preprocessing.

```python
sorted_df = df.sort_values("Marks", ascending=False)

print(sorted_df)
```

### Explanation

The `sort_values()` method sorts the DataFrame using the Marks column.

The `ascending=False` argument arranges the marks from highest to lowest.

### Best Practices

Specify the correct column and sorting order according to the requirement.

### Common Mistakes

Forgetting that `ascending=False` sorts values in descending order, while `ascending=True` sorts them in ascending order.
