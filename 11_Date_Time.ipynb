#  Module  11 – Date & Time Operations

## 1. Datetime

Datetime represents both a date and a time in a single value.

It helps us store, compare, and analyze time-based information in Pandas.

### Real-World Example

A student's attendance can be recorded with both the date and time of arrival.

### Business Example

An online store can record the exact date and time when an order is placed.

### AI/ML Example

Datetime values can be used to analyze trends over time and prepare time-based features for Machine Learning.

```python
import pandas as pd

dt = pd.to_datetime("2026-08-20 10:30:00")

print(dt)
```

### Explanation

The `to_datetime()` function converts a text value into a Pandas datetime object.

The result contains both the date and the time.

### Best Practices

Store time-related data as datetime objects instead of plain text.

### Common Mistakes

Keeping dates as strings makes time-based analysis more difficult.

## 2. Timestamp

A Timestamp represents a specific date and time.

In Pandas, timestamps are useful for working with individual points in time and performing date and time operations.

### Real-World Example

A student's exam result can be recorded with the exact date and time when the result was published.

### Business Example

An online transaction can be recorded using a timestamp to identify exactly when the transaction occurred.

### AI/ML Example

Timestamps can be used to analyze time-based events and create useful time-related features for Machine Learning.

```python
import pandas as pd

timestamp = pd.Timestamp("2026-08-20 10:30:00")

print(timestamp)
print(timestamp.year)
print(timestamp.month)
print(timestamp.day)
```

### Explanation

`pd.Timestamp()` creates a Pandas Timestamp representing a specific date and time.

We can also access individual parts of the timestamp, such as the year, month, and day.

### Best Practices

Use Pandas Timestamp when working with individual date and time values.

### Common Mistakes

Using an invalid date or incorrect date format can cause an error.

## 3. Date Parsing

Date Parsing is the process of converting date values from text or other formats into a proper date and time format.

In Pandas, `pd.to_datetime()` is commonly used to convert date values into datetime objects.

### Real-World Example

A student's birth date may be stored as text and can be converted into a proper date format for further analysis.

### Business Example

A company may receive order dates in different formats and convert them into a consistent datetime format for sales analysis.

### AI/ML Example

During data preprocessing, date values can be converted into datetime format so that useful features such as year, month, and day can be extracted for Machine Learning.

```python
import pandas as pd

dates = pd.Series([
    "2026-08-20",
    "2026-09-15",
    "2026-10-05"
])

parsed_dates = pd.to_datetime(dates)

print(parsed_dates)
```

### Explanation

The `pd.to_datetime()` function converts the date values stored as text into Pandas datetime values.

After conversion, the dates can be used for operations such as sorting, filtering, calculating differences, and extracting date features.

### Best Practices

Convert date columns to datetime format before performing date and time analysis.

### Common Mistakes

Inconsistent or invalid date values can cause parsing errors or unexpected results.

## 4. Timedelta

`Timedelta` represents a difference or duration between two dates or times.

It is useful for calculating how much time has passed between two datetime values.

### Real-World Example

We can calculate how many days a student attended classes or how long an event lasted.

### Business Example

A company can calculate the number of days between an order date and delivery date.

### AI/ML Example

During data preprocessing, time differences can be used to create useful features such as delivery duration, customer age, or time between transactions.

```python
import pandas as pd

start_date = pd.Timestamp("2026-08-20")
end_date = pd.Timestamp("2026-08-25")

duration = end_date - start_date

print(duration)
print("Days:", duration.days)
```

### Explanation

The difference between the `end_date` and `start_date` is calculated as a Timedelta.

The result shows that 5 days have passed between the two dates.

### Best Practices

Use Timedelta when calculating durations or differences between dates and times.

### Common Mistakes

Subtracting values that are not in a proper datetime format can produce errors or incorrect results.

## 5. Time Zones

A Time Zone represents the local time of a particular geographical region.

Pandas can work with time zones to convert and compare datetime values from different locations.

### Real-World Example

A meeting scheduled in India may need to be converted to the local time of someone working in the United States.

### Business Example

A global company can use time zones to manage transactions, meetings, and customer activities across different countries.

### AI/ML Example

Time zone information can be useful when analyzing time-based data collected from users or systems located in different regions.

```python
import pandas as pd

date = pd.Timestamp("2026-08-20 10:00:00", tz="Asia/Kolkata")

print(date)
print(date.tz_convert("UTC"))
```

### Explanation

The timestamp is first created using the `Asia/Kolkata` time zone.

The `tz_convert()` method converts the same moment to another time zone, which is UTC in this example.

### Best Practices

Use a consistent time zone when storing and processing time-based data.

### Common Mistakes

Confusing a time zone conversion with changing the actual moment in time. Conversion changes the displayed local time, not the event itself.

## 6. Extracting Date Features

Extracting Date Features means taking useful information such as year, month, day, day of the week, and hour from a datetime value.

These features help us analyze time-based data and understand patterns over different time periods.

### Real-World Example

We can extract the day, month, or year from a student's attendance date to understand attendance patterns.

### Business Example

A company can extract the month or day of the week from order dates to identify when sales are higher.

### AI/ML Example

Date features such as year, month, day, and weekday can be used as input features when building Machine Learning models for time-based prediction.

```python
import pandas as pd

date = pd.Timestamp("2026-08-20 10:30:00")

print("Year:", date.year)
print("Month:", date.month)
print("Day:", date.day)
print("Day of Week:", date.day_name())
```

### Explanation

The datetime value contains different pieces of information such as year, month, day, and day of the week.

These values can be extracted individually and used for further analysis.

### Best Practices

Extract only the date features that are relevant to the analysis or Machine Learning task.

### Common Mistakes

Using date features without first converting the column into a proper datetime format.
