## 1. What is Pandas?

Pandas is a popular Python library used for working with structured and tabular data. It provides easy-to-use data structures such as Series and DataFrame for storing, analyzing, and manipulating data.

Pandas is commonly used for tasks such as loading datasets, inspecting data, cleaning missing values, filtering records, and preparing data for analysis and Machine Learning.

### Real-World Example

Consider a student dataset containing names, ages, marks, and grades. Pandas can organize this information into a table and make it easy to search, filter, and analyze student records.

### Business Example

A company may have thousands of customer records containing customer names, locations, purchases, and revenue. Pandas can be used to analyze this data and find useful information such as total sales, average purchase value, and customer trends.

### AI/ML Example

Before training a Machine Learning model, data usually needs to be loaded, inspected, cleaned, and prepared. Pandas helps perform these tasks efficiently and prepares the dataset for further Machine Learning operations.

```python
import pandas as pd

data = {
    "Name": ["Saiprakash", "Neha", "Sampradaya"],
    "Age": [23, 22, 20],
    "Marks": [85, 78, 92]
}

df = pd.DataFrame(data)

print(df)
```

###  Explanation

The `pd.DataFrame()` function converts the dictionary into a table.

The DataFrame contains three columns: `Name`, `Age`, and `Marks`.

Pandas automatically creates an index starting from 0.

### Best Practices

Use meaningful column names and keep data organized when creating a DataFrame.

### Common Mistakes

Forgetting to import Pandas or providing different numbers of values for different columns.

## 2. Why Pandas?

Pandas makes it easier to work with structured and tabular data in Python.

It provides simple functions for loading, viewing, filtering, cleaning, and analyzing datasets.

Pandas is especially useful when working with large datasets because it provides many operations for data manipulation and analysis.


## 3. Series

A Pandas Series is a one-dimensional data structure used to store a sequence of values.

Each value in a Series has an index that helps identify and access individual values. A Series can contain numerical, text, or other types of data.

A Series can be considered as a single column of data.

### Real-World Example

Student marks can be stored in a Series where each value represents the marks obtained by a student.

### Business Example

A company can store monthly sales values for a product in a Pandas Series and use them to calculate totals, averages, or trends.

### AI/ML Example

Individual columns of a dataset are commonly represented as Series in Pandas. Series can be used during data inspection, analysis, and preprocessing.

```python
marks = pd.Series([85, 78, 92])

print(marks)
```

###  Explanation

The `pd.Series()` function creates a one-dimensional Series from the given values.

Pandas automatically assigns index numbers starting from 0.

The values in the Series are stored as integers.

### Best Practices

Use a Series when working with one-dimensional data or a single column.

### Common Mistakes

Confusing a Series with a DataFrame. A Series is one-dimensional, while a DataFrame is two-dimensional.

## 4. DataFrame

A Pandas DataFrame is a two-dimensional data structure that stores data in rows and columns.

It is similar to a table or spreadsheet and can contain multiple columns with different data types.

A DataFrame is one of the most commonly used Pandas structures for data analysis.

### Real-World Example

Student information such as name, age, and marks can be organized into a DataFrame where each row represents a student and each column represents a type of information.

### Business Example

A company can use a DataFrame to store customer information such as customer name, age, location, purchase amount, and product details.

### AI/ML Example

Most structured datasets used in Machine Learning can be represented using Pandas DataFrames. They can be inspected, cleaned, transformed, and prepared before model training.

```python
data = {
    "Name": ["Saiprakash", "Neha", "Sampradaya"],
    "Age": [23, 22, 20],
    "Marks": [85, 78, 92]
}

df = pd.DataFrame(data)

print(df)
```

###  Explanation

The `pd.DataFrame()` function converts the dictionary into a tabular DataFrame.

The dictionary keys become the column names.

Each row represents one student and the index starts from 0.

### Best Practices

Use meaningful column names and keep related information in separate columns.

### Common Mistakes

Providing different numbers of values for different columns when creating a DataFrame.

## 5. Creating DataFrames

A DataFrame can be created from different Python data structures such as dictionaries, lists, and NumPy arrays.

A dictionary is commonly used because the dictionary keys can become the column names and the values become the data.

Creating DataFrames is useful when we have data in Python and want to organize it into a tabular structure.

### Real-World Example

Student records stored as Python lists can be converted into a DataFrame to make the information easier to analyze.

### Business Example

Customer information collected from an application can be stored in Python structures and converted into a DataFrame for further analysis.

### AI/ML Example

Data collected during data preparation can be converted into a DataFrame before performing cleaning, feature engineering, and other Machine Learning preprocessing tasks.

```python
data = {
    "Name": ["Saiprakash", "Neha", "Sampradaya"],
    "Age": [23, 22, 20],
    "Marks": [85, 78, 92]
}

df = pd.DataFrame(data)

print(df)
```

###  Explanation

The dictionary contains three columns: Name, Age, and Marks.

The `pd.DataFrame()` function converts this dictionary into a DataFrame.

### Best Practices

Make sure the column names are meaningful and each column contains the correct type of information.

### Common Mistakes

Using different numbers of values in different columns.

## 6. Data Types

Data types describe the type of values stored in each column of a DataFrame.

Pandas supports different data types such as integers, floating-point numbers, strings, and boolean values.

Understanding data types is important because different operations can be performed depending on the type of data.

### Real-World Example

A student's age can be stored as an integer, while the student's name can be stored as text.

### Business Example

A sales dataset may contain product names as text, quantities as integers, and prices as floating-point numbers.

### AI/ML Example

Checking data types is an important step during Machine Learning preprocessing because models generally require numerical features in a suitable format.

```python
data = {
    "Name": ["Saiprakash", "Neha", "Sampradaya"],
    "Age": [23, 22, 20],
    "Marks": [85, 78, 92]
}

df = pd.DataFrame(data)

print(df.dtypes)
```

###  Explanation

The `dtypes` attribute displays the data type of each column.

The Name column contains text, while Age and Marks contain integer values.

### Best Practices

Check the data types of columns before performing calculations or preprocessing.

### Common Mistakes

Assuming that every column contains numerical data.

## 7. Index

An index is used to identify the rows of a DataFrame or the values of a Series.

By default, Pandas creates a numerical index starting from 0.

Indexes help us access and identify specific records in a dataset.

### Real-World Example

Each student record in a student table can be identified using its row index.

### Business Example

A company can use an index to identify individual customer or transaction records.

### AI/ML Example

Indexes are useful when selecting, filtering, and accessing specific rows during data analysis and preprocessing.

```python
data = {
    "Name": ["Saiprakash", "Neha", "Sampradaya"],
    "Marks": [85, 78, 92]
}

df = pd.DataFrame(data)

print(df.index)
```

### Explanation

The DataFrame contains three rows.

The index starts at 0 and continues up to 2.

`RangeIndex` represents the default numerical index created by Pandas.

### Best Practices

Use meaningful indexes when they help identify records.

### Common Mistakes

Forgetting that Pandas uses zero-based indexing by default.

## 8. Columns

Columns represent the individual fields or variables in a DataFrame.

Each column usually stores one type of information, such as a name, age, marks, price, or quantity.

Columns are important because they allow us to select and analyze specific parts of a dataset.

### Real-World Example

A student dataset can contain columns such as Name, Age, and Marks.

### Business Example

A sales dataset can contain columns such as Product, Quantity, Price, and Revenue.

### AI/ML Example

Columns can represent features that are used as inputs when building Machine Learning models.

```python
data = {
    "Name": ["Saiprakash", "Neha", "Sampradaya"],
    "Age": [23, 22, 20],
    "Marks": [85, 78, 92]
}

df = pd.DataFrame(data)

print(df.columns)
```

###  Explanation

The `columns` attribute returns the names of all columns in the DataFrame.

The DataFrame contains three columns: Name, Age, and Marks.

### Best Practices

Use clear, descriptive, and consistent column names.

### Common Mistakes

Using an incorrect column name while selecting or processing data.

## 9. Reading Data

Reading data means loading information from an external file into a Pandas DataFrame.

Pandas provides functions such as `read_csv()` to read CSV files and convert their contents into a DataFrame.

Reading data is one of the first steps in a data analysis workflow.

### Real-World Example

A student dataset stored in a CSV file can be loaded into Pandas and analyzed.

### Business Example

A company can load customer or sales records stored in CSV files and use Pandas to analyze the information.

### AI/ML Example

Machine Learning datasets are often stored in files. Pandas can load these datasets into DataFrames before data cleaning and preprocessing.

```python
import pandas as pd

data = {
    "Name": ["Saiprakash", "Neha", "Sampradaya"],
    "Age": [23, 22, 20],
    "Marks": [85, 78, 92]
}

df = pd.DataFrame(data)

df.to_csv("students.csv", index=False)

print("CSV file created successfully")
```

```python
df = pd.read_csv("students.csv")

print(df)
```

## 10. Writing Data

Writing data means saving a Pandas DataFrame to an external file.

Pandas provides functions such as `to_csv()` to save DataFrames as CSV files.

Saving data is useful when we want to store processed information and use it later.

### Real-World Example

Student information can be saved as a CSV file so that it can be opened and analyzed later.

### Business Example

A company can save processed sales or customer data into a CSV file and share it with another team.

### AI/ML Example

After cleaning and preprocessing a dataset, the processed data can be saved and reused for Machine Learning experiments.

```python
data = {
    "Name": ["Saiprakash", "Neha", "Sampradaya"],
    "Age": [23, 22, 20],
    "Marks": [85, 78, 92]
}

df = pd.DataFrame(data)

df.to_csv("students.csv", index=False)

print("Data saved successfully")
```

###  Explanation

The `to_csv()` function saves the DataFrame as a CSV file named `students.csv`.

The `index=False` parameter prevents the DataFrame index from being saved as an additional column.

### Best Practices

Use `index=False` when the DataFrame index is not required in the saved file.

### Common Mistakes

Using an incorrect file path or forgetting to specify the correct file name.
