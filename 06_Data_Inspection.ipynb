## 1. head()

The `head()` method is used to display the first few rows of a Pandas DataFrame.

By default, `head()` displays the first 5 rows. We can also provide a number to display a specific number of rows.

### Real-World Example

A student dataset may contain many records. `head()` helps us quickly view the first few records.

### Business Example

A company can use `head()` to quickly check the first few customer or sales records after loading a dataset.

### AI/ML Example

Before starting data preprocessing, `head()` helps us understand the structure and values of the dataset.

```python
import pandas as pd

data = {
    "Name": ["Sai", "Rahul", "Priya"],
    "Age": [23, 22, 20],
    "Marks": [85, 78, 92]
}

df = pd.DataFrame(data)

print(df.head())
```

###  Explanation

The `head()` method displays the first few rows of the DataFrame.

Since our DataFrame contains only three rows, all three rows are displayed.

### Best Practices

Use `head()` after loading a dataset to quickly check the data.

### Common Mistakes

Assuming `head()` displays the complete dataset. It only displays the first few rows.

## 2. tail()

The `tail()` method is used to display the last few rows of a Pandas DataFrame.

By default, `tail()` displays the last 5 rows. We can also provide a number to display a specific number of rows.

### Real-World Example

A student dataset may contain many records. `tail()` helps us quickly check the last few student records.

### Business Example

A company can use `tail()` to check the most recent records in a sales or customer dataset.

### AI/ML Example

`tail()` can be used to check the end of a dataset and verify whether the data was loaded completely.


```python
print(df.tail())
```

###  Explanation

The `tail()` method displays the last few rows of the DataFrame.

Since the DataFrame contains only three rows, all three rows are displayed.

### Best Practices

Use `tail()` along with `head()` to inspect both the beginning and end of a dataset.

### Common Mistakes

Confusing `tail()` with `head()`. `tail()` displays the last rows, not the first rows.

## 3. sample()

The `sample()` method is used to randomly select rows from a Pandas DataFrame.

It is useful when we want to quickly look at random records instead of always viewing the first or last rows.

### Real-World Example

A large student dataset can contain thousands of records. `sample()` can randomly select a few students for inspection.

### Business Example

A company can randomly inspect customer or transaction records using `sample()`.

### AI/ML Example

Random samples can help us understand whether the dataset contains different types of records before preprocessing.


```python
print(df.sample(2, random_state=1))
```

###  Explanation

The `sample(2)` method selects two random rows from the DataFrame.

The `random_state=1` makes the random selection reproducible.

### Best Practices

Use `tail()` along with `head()` to inspect both the beginning and end of a dataset.

### Common Mistakes

Confusing `tail()` with `head()`. `tail()` displays the last rows, not the first rows.

## 4. info()

The `info()` method provides a summary of a Pandas DataFrame.

It shows information such as the number of rows, columns, column names, non-null values, and data types.

### Real-World Example

After loading a dataset, `info()` can help us quickly understand how many records and columns are present.

### Business Example

A company can use `info()` to check the structure of customer or sales data before analysis.

### AI/ML Example

`info()` is useful during data preprocessing because it helps identify missing values and understand column data types.



```python
df.info()
```

### Explanation

The `info()` method shows that the DataFrame contains 3 rows and 3 columns.

It also shows the number of non-null values and the data type of each column.

### Best Practices

Use `info()` early when inspecting a newly loaded dataset.

### Common Mistakes

Looking only at the number of rows and ignoring the data types or non-null counts.

## 5. describe()

The `describe()` method provides statistical information about numerical columns in a DataFrame.

It commonly displays count, mean, standard deviation, minimum, quartiles, and maximum values.

### Real-World Example

A teacher can use `describe()` to understand the overall performance of students based on their marks.

### Business Example

A company can use `describe()` to understand sales values such as average, minimum, and maximum sales.

### AI/ML Example

`describe()` helps understand numerical features before Machine Learning preprocessing.



```python
print(df.describe())
```

###  Explanation

The `describe()` method provides statistical information for the Age and Marks columns.

It shows values such as mean, minimum, maximum, and quartiles.


### Best Practices

Use `describe()` to quickly understand the distribution of numerical data.

### Common Mistakes

Assuming `describe()` provides the same statistical summary for text columns by default.


## 6. shape

The `shape` attribute returns the number of rows and columns in a DataFrame.

It returns the result in the form `(rows, columns)`.

### Real-World Example

A dataset can contain thousands of rows and many columns. `shape` quickly tells us the size of the dataset.

### Business Example

A company can use `shape` to check how many customer records and fields are available.

### AI/ML Example

`shape` helps understand the number of observations and features before building a Machine Learning model.



```python
print(df.shape)
```

###  Explanation

The output `(3, 3)` means the DataFrame contains 3 rows and 3 columns.

### Best Practices

Check the shape of a dataset after loading it.

### Common Mistakes

Confusing the order of the values. The first value represents rows and the second represents columns.

## 7. columns

The `columns` attribute returns the names of all columns in a Pandas DataFrame.

It is useful when we want to understand what variables are available in a dataset.

### Real-World Example

A student dataset may contain columns such as Name, Age, and Marks.

### Business Example

A sales dataset may contain columns such as Product, Quantity, Price, and Revenue.

### AI/ML Example

Column names help identify the features available for Machine Learning.



```python
print(df.columns)
```

###  Explanation

The `columns` attribute returns the names of all three columns in the DataFrame.
### Best Practices

Check column names before selecting or processing specific columns.

### Common Mistakes

Using a column name that does not exist in the DataFrame.

## 8. dtypes

The `dtypes` attribute displays the data type of each column in a DataFrame.

Understanding data types is important before performing calculations or data preprocessing.

### Real-World Example

A student's name can be stored as text, while age and marks can be stored as integers.

### Business Example

Product names may be text, quantities may be integers, and prices may be floating-point numbers.

### AI/ML Example

Checking data types helps identify which columns are numerical and which columns may need conversion before Machine Learning.



```python
print(df.dtypes)
```

###  Explanation

The `dtypes` attribute shows that Name contains text data, while Age and Marks contain integer data.
### Best Practices

Check data types before performing calculations or preprocessing.

### Common Mistakes

Assuming that every column in a dataset contains numerical data.

## 9. memory_usage()

The `memory_usage()` method shows the amount of memory used by each column of a DataFrame.

It is useful when working with large datasets and checking memory consumption.

### Real-World Example

A large dataset with millions of records may require a significant amount of memory. `memory_usage()` helps inspect memory usage.

### Business Example

A company working with large customer or transaction datasets can use memory information to manage data efficiently.

### AI/ML Example

Memory usage is important when preparing large datasets for Machine Learning because inefficient data can consume more system resources.



```python
print(df.memory_usage())
```

### Explanation

The `memory_usage()` method shows the memory used by the DataFrame index and each column.

The values are shown in bytes.

### Best Practices

Check memory usage when working with large datasets.

### Common Mistakes

Ignoring memory usage when working with very large DataFrames

## 10. nunique()

The `nunique()` method returns the number of unique values in each column of a DataFrame.

It is useful for understanding how many different values are present in each column.

### Real-World Example

A student dataset can be checked to find the number of unique names, ages, or marks.

### Business Example

A company can use `nunique()` to find the number of unique products, customers, or locations.

### AI/ML Example

`nunique()` can help identify categorical features and understand the number of unique values before preprocessing.



```python
print(df.nunique())
```

###  Explanation

The `nunique()` method counts the number of different values in each column.

In this DataFrame, each column contains three unique values.

### Best Practices

Use `nunique()` to understand the uniqueness of values in different columns.

### Common Mistakes

Confusing the number of unique values with the total number of rows.

## 11. value_counts()

The `value_counts()` method counts how many times each unique value appears in a Series or DataFrame column.

It is useful for understanding the frequency of values in categorical or repeated data.

### Real-World Example

A teacher can use `value_counts()` to count how many students received each grade.

### Business Example

A company can use it to count how many customers belong to each city or how many products were sold in each category.

### AI/ML Example

`value_counts()` can help inspect categorical features and identify the distribution of classes in a dataset.



```python
data = {
    "Name": ["Sai", "Rahul", "Priya", "Amit"],
    "Age": [23, 22, 20, 21],
    "Marks": [85, 78, 85, 92]
}

df = pd.DataFrame(data)

print(df["Marks"].value_counts())
```

###  Explanation

The `value_counts()` method counts how many times each mark appears.

The value 85 appears two times, while 78 and 92 appear one time each.

### Best Practices

Use `nunique()` to understand the uniqueness of values in different columns.

### Common Mistakes

Confusing the number of unique values with the total number of rows.

## Advantages

- Data Inspection methods make it easier to understand a dataset.
- They provide quick information about rows, columns, data types, and values.
- Statistical summaries can be obtained using simple Pandas methods.
- They help identify possible data quality issues.
- They are useful during Data Science and Machine Learning preprocessing.

## Limitations

- Inspection methods provide information about the data but do not automatically clean the data.
- Large datasets may require more memory and processing time.
- Some methods provide summaries that may not show the complete dataset.
- Statistical summaries mainly focus on numerical data.
